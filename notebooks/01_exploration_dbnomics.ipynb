```python
import requests

r = requests.get("https://api.db.nomics.world/v22/series/BCEAO/TAUXINFL_A?observations=0&limit=1000")
data = r.json()

for s in data['series']['docs']:
    print(s['series_code'], '-', s['series_name'])
```

```python
# On parcourt tous les codes de séries qu'on a récupérés dans "data"
for s in data['series']['docs']:
    # On vérifie si "SENEGAL" apparaît dans le nom de la série (en majuscules, pour éviter les soucis de casse)
    if 'SENEGAL' in s['series_name'].upper():
        # Si oui, on affiche le code de la série et son nom complet
        print(s['series_code'], '-', s['series_name'])
```

```python
# On récupère la série du taux d'inflation moyen annuel (IPC) pour le Sénégal
from dbnomics import fetch_series

df = fetch_series('BCEAO/TAUXINFL_A/KKKSR3072A0BP')

# On affiche les 5 premières lignes pour vérifier que ça a bien fonctionné
df.head()
```

```python
# On affiche les dernières lignes pour voir si les données récentes existent
print(df.tail())

# On compte combien de valeurs ne sont pas vides (NaN)
print("Nombre de valeurs non-vides :", df['value'].notna().sum())

# On affiche la première et la dernière année où on a une vraie valeur
donnees_valides = df[df['value'].notna()]
print("Première année avec donnée :", donnees_valides['period'].min())
print("Dernière année avec donnée :", donnees_valides['period'].max())
```

```python
# Côte d'Ivoire
from dbnomics import fetch_series
df_inflation_ci = fetch_series('BCEAO/TAUXINFL_A/AAASR3072A0BP')
df_pib_ci = fetch_series('BCEAO/PIBN/AAASR1015A0BP')

# Burkina Faso
df_inflation_bf = fetch_series('BCEAO/TAUXINFL_A/CCCSR3072A0BP')
df_pib_bf = fetch_series('BCEAO/PIBN/CCCSR1015A0BP')

# Mali
df_inflation_mali = fetch_series('BCEAO/TAUXINFL_A/DDDSR3072A0BP')
df_pib_mali = fetch_series('BCEAO/PIBN/DDDSR1015A0BP')

print(df_pib_ci.shape, df_pib_bf.shape, df_pib_mali.shape)
print(df_inflation_ci.shape, df_inflation_bf.shape, df_inflation_mali.shape)
```

```python
import pathlib
dossier_raw = pathlib.Path("../data/raw")

df_pib_ci.to_csv(dossier_raw / "pib_cote_ivoire.csv", index=False)
df_inflation_ci.to_csv(dossier_raw / "inflation_cote_ivoire.csv", index=False)
df_pib_bf.to_csv(dossier_raw / "pib_burkina.csv", index=False)
df_inflation_bf.to_csv(dossier_raw / "inflation_burkina.csv", index=False)
df_pib_mali.to_csv(dossier_raw / "pib_mali.csv", index=False)
df_inflation_mali.to_csv(dossier_raw / "inflation_mali.csv", index=False)

print("6 fichiers sauvegardés dans data/raw/")
```
